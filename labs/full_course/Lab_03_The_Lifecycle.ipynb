# Lab 03: The Lifecycle (Environment Setup)
Validating the actual open-source pipeline (Ollama, SCALE-Sim, Yosys) in your environment.


```
import shutil
import json

tools = ['yosys', 'python3', 'ollama']
pipeline_ready = True
for tool in tools:
    path = shutil.which(tool)
    if path:
        print(f"✅ {tool} found at: {path}")
    else:
        print(f"❌ {tool} not found!")
        pipeline_ready = False

with open('.arch2_state.json', 'r') as f: state = json.load(f)
state['pipeline_ready'] = pipeline_ready
with open('.arch2_state.json', 'w') as f: json.dump(state, f)
```
