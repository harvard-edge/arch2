# Lab 06: Environments & IP Isolation

IP Isolation is more than just access control; it requires strict legal and state boundaries preventing the contamination of LLM context windows with proprietary PDK data.

## Simulating Context Leakage
If an LLM retains its context across a tool boundary, the generated output legally compromises the proprietary physical verification data.

```
class GenerativeLLM:
    def __init__(self):
        self.context = ["GPLv3_ALU_Code", "OpenCores_DMA"]
    
    def generate(self, prompt):
        return f"Generated based on: {self.context} + {prompt}"

# The LLM leaks its open-source context into the proprietary domain
llm = GenerativeLLM()
proprietary_pdk_rules = "TSMC_N3_FinFET_Rules"

# A secure environment MUST wipe the LLM context here
# llm.context.clear() # Without this, we have contamination!

llm.context.append(proprietary_pdk_rules)
output = llm.generate("Optimize MAC for Area")

if "GPLv3" in output and "TSMC" in output:
    print("🚨 CRITICAL IP CONTAMINATION DETECTED 🚨")
    print(output)
else:
    print("✅ Isolation Successful")
```
