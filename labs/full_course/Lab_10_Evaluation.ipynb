# Lab 10: Evaluation (Pareto Frontiers)
Plotting Pareto frontiers based on the synthesis results.


```
import matplotlib.pyplot as plt
import json

with open('.arch2_state.json', 'r') as f: state = json.load(f)
proxy_cycles = state.get('proxy_cycles', 100000)
cells = state.get('yosys_cells', 500)

latencies = [400000, 200000, 80000, proxy_cycles]
areas = [100, 250, 800, cells]
labels = ['LLM Config A', 'LLM Config B', 'LLM Config C', 'Your LLM Run']

plt.scatter(latencies, areas, c=['blue','blue','blue','red'], s=100)
for i, label in enumerate(labels):
    plt.annotate(label, (latencies[i]+5000, areas[i]))

plt.xlabel('Latency (Cycles)')
plt.ylabel('Area (Standard Cells)')
plt.title('Area vs Latency Tradeoff (Pareto Frontier)')
plt.grid(True)
plt.show()
```
