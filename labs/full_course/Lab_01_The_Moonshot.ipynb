# Lab 01: The Moonshot (IP & Provenance)
We inspect an open-source hardware dataset to identify licenses and understand the legal risks of raw LLM regurgitation.


```
import pandas as pd
import json

data = {
    'module': ['ALU', 'Multiplier', 'Cache_Ctrl', 'DMA_Engine'],
    'source': ['OpenCores', 'GitHub_UserX', 'Internal', 'Unknown_LLM_Output'],
    'license': ['GPLv3', 'MIT', 'Proprietary', 'None']
}
df = pd.DataFrame(data)

def check_license_risk(row):
    if row['license'] == 'GPLv3': return 'HIGH RISK (Viral)'
    elif row['license'] == 'None': return 'CRITICAL RISK (Unknown)'
    return 'SAFE'

df['Risk Level'] = df.apply(check_license_risk, axis=1)
print(df)

# Initialize our persistent state for the workshop
state = {'lab01_complete': True}
with open('.arch2_state.json', 'w') as f:
    json.dump(state, f)
```
