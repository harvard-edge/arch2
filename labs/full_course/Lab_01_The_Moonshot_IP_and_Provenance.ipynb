# Lab 01: The Moonshot

In Chapter 1, we introduced the concept of IP contamination and the risk of raw LLM regurgitation. When AI models generate hardware code based on open-source datasets (like GPLv3), they can inadvertently poison proprietary workflows.

### Theoretical Background
The transition from software SWE-bench to hardware benchmarking requires strict IP boundary management. Hardware tapeouts cost millions of dollars, and a single GPL-tainted module generated by an LLM can void the legal viability of an entire System-on-Chip (SoC).

## Exercise: Interactive IP Risk Evaluation
Use the form below to input a generated hardware module, its source, and license. The system will evaluate its legal risk profile.

```
#@title Interactive IP Risk Evaluator
module_name = 'DMA_Engine' #@param {type:"string"}
source = 'Unknown_LLM_Output' #@param ["OpenCores", "GitHub_UserX", "Internal", "Unknown_LLM_Output"]
license_type = 'None' #@param ["GPLv3", "MIT", "Proprietary", "None"]

def evaluate_risk(lic):
    if lic == 'GPLv3': return 'HIGH RISK (Viral)'
    elif lic == 'None': return 'CRITICAL RISK (Unknown Provenance/Poisoning)'
    return 'SAFE'

print(f"Evaluating {module_name} from {source}...")
print(f"Risk Level: {evaluate_risk(license_type)}")
```

### Visualizing the Contamination Problem
Let's visualize the ratio of safe vs. contaminated modules in a synthetic dataset of 1,000 LLM-generated Verilog files.

```
import matplotlib.pyplot as plt
import pandas as pd

data = [600, 250, 150]
labels = ['Safe (MIT/Apache)', 'Viral (GPL)', 'Unknown (Critical)']
colors = ['#4CAF50', '#FF9800', '#F44336']

plt.figure(figsize=(6,6))
plt.pie(data, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)
plt.title('IP Profile of LLM-Generated Hardware Dataset')
plt.show()
```
