Chapter 2 highlights the tension between open-source ecosystems and proprietary 'Data Moats'.

### The Danger of the Data Moat
A Data Moat represents the fidelity gap between open-source proxies (like SCALE-Sim) and closely guarded physical PPA outcomes (like TSMC N3 wire delays). Open-source tools might predict a design is optimal, but when that design hits the proprietary physical synthesis tools, it fails catastrophically due to non-linear physical realities (e.g., routing congestion).

## Exercise: Simulating the Physical Reality Gap
Use the sliders below to adjust the complexity of your design (Gate Count). Notice how the open-source estimator always returns a valid answer, but the proprietary physical validation fails when physical limits are exceeded.

```
import ipywidgets as widgets
from IPython.display import display

#@title Design Complexity Form
gates = 1500000 #@param {type:"slider", min:100000, max:5000000, step:100000}

class OpenSourceEstimator:
    def estimate_delay(self, gates):
        return (gates / 100000) * 1.5 # Linear, naive assumption

class ProprietaryPDK:
    def __init__(self):
        self.wire_delay_wall = 2500000 # The hidden physical limit
    
    def validate(self, gates):
        # Non-linear wire congestion delay
        if gates > self.wire_delay_wall:
            raise ValueError(f"Physical Routing Failure: Congestion exploded at {gates} gates!")
        return (gates / 100000) * 1.6

open_tool = OpenSourceEstimator()
foundry = ProprietaryPDK()

print(f"--- Open Source Estimate ---")
print(f"Predicted Delay: {open_tool.estimate_delay(gates):.2f} ns\n")

print(f"--- Proprietary Physical Validation ---")
try:
    actual = foundry.validate(gates)
    print(f"Actual Delay: {actual:.2f} ns (Success!)")
except Exception as e:
    print(f"❌ ERROR: {e}")
```
