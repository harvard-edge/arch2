# Lab 03: The XR Lighthouse Request

In Chapter 3, the XR Lighthouse is defined as a comprehensive "SWE-bench for hardware"—a 64-bit RISC-V core operating within a 3W thermal envelope with complex non-linear trade-offs.

## Exercise: Non-Linear Trade-offs
A static dictionary is not a benchmark. Here, we simulate a parameterized hardware model where choices have secondary effects. Shrinking the L2 cache reduces Area, but increases off-chip traffic, which spikes Power and degrades Latency.

```
#@title Configure the XR Subsystem
l2_cache_size_kb = 128 #@param {type:"slider", min:64, max:1024, step:64}

def simulate_xr_workload(cache_kb):
    # Base metrics
    area = 2.0 + (cache_kb * 0.01)
    
    # Non-linear effects: smaller cache = huge off-chip traffic
    miss_rate = 1.0 / (cache_kb ** 0.5)
    off_chip_traffic_penalty = miss_rate * 50
    
    power_w = 1.5 + (cache_kb * 0.002) + (off_chip_traffic_penalty * 0.1)
    latency_ms = 8 + off_chip_traffic_penalty
    
    return {'Area (mm2)': area, 'Power (W)': power_w, 'Latency (ms)': latency_ms}

results = simulate_xr_workload(l2_cache_size_kb)
print(f"Testing XR Lighthouse with {l2_cache_size_kb}KB L2 Cache:\n")
for k, v in results.items():
    status = '✅' if (k == 'Power (W)' and v < 3.0) or (k == 'Latency (ms)' and v < 15) else '⚠️'
    print(f"{k:15}: {v:.2f} {status}")
```
