# Lab 09: Patterns (Design Scalability)
Generating heavily parameterized Verilog templates in Python so AI acts as an orchestrator, not a raw coder.


```
def generate_parameterized_mac(w, d):
    return f"""module smart_mac #(parameter W={w}, D={d}) (
    input clk, input [W-1:0] a, input [W-1:0] b, output reg [2*W-1:0] out
);
    always @(posedge clk) out <= a * b;
endmodule
"""

with open('smart_mac.v', 'w') as f: f.write(generate_parameterized_mac(16, 8))
print("Parameterized smart_mac.v generated.")
```
