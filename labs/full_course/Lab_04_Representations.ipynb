# Lab 04: Representations (The Lighthouse Request)
Formalizing product intent into rigid physical constraints via a YAML Loop Card.


```
%%writefile loop_card.yaml
product: XR_SoC_Gen3
power_budget_w: 3.0
constraints:
  max_cycles: 500000
  min_pe_utilization: 60.0
architecture:
  type: systolic_array
```

```
import yaml
import json

with open('loop_card.yaml', 'r') as f:
    card = yaml.safe_load(f)

print(f"Validating Loop Card for {card['product']}...")
assert card['power_budget_w'] <= 3.0
max_cyc = card['constraints']['max_cycles']

with open('.arch2_state.json', 'r') as f: state = json.load(f)
state['max_cycles'] = max_cyc
with open('.arch2_state.json', 'w') as f: json.dump(state, f)
print("Constraints saved to state.")
```
