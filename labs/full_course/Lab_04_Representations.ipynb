# Lab 04: Representations

Translating intent into machine-readable JSON representations.

```
import json

architecture_representation = {
    'version': '1.0',
    'subsystem': 'Vision_Accelerator',
    'components': [
        {'type': 'MAC_Array', 'size': '16x16'},
        {'type': 'SRAM_Buffer', 'size_kb': 256}
    ]
}
print(json.dumps(architecture_representation, indent=2))
```
