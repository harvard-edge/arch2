## Environment Check
Let's verify that your local LLM (Ollama) and EDA tools are installed.

*Troubleshooting:* 
* If Ollama fails, run `ollama serve` in a new terminal.
* If `gemma3:1b` is missing, run `ollama pull gemma3:1b`.
* If Yosys fails, run `brew install yosys` (Mac) or `sudo apt install yosys` (Linux).


````
import subprocess
import urllib.request
import json

print("Checking Tools...")
# Check Ollama and Gemma3:1b
try:
    req = urllib.request.Request("http://localhost:11434/api/tags")
    res = urllib.request.urlopen(req)
    models = [m['name'] for m in json.loads(res.read())['models']]
    if any('gemma3:1b' in m for m in models):
        print("✅ Ollama is running and gemma3:1b is installed.")
    else:
        print("❌ Ollama is running, but gemma3:1b is NOT installed. Run `ollama pull gemma3:1b`.")
except Exception:
    print("❌ Ollama is not responding. Please start the Ollama daemon.")

# Check Yosys
try:
    subprocess.run(["yosys", "-V"], check=True, capture_output=True)
    print("✅ Yosys logic synthesizer is installed.")
except Exception:
    print("❌ Yosys not found.")
````
