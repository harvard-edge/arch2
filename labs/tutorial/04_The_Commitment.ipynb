## The Threat of IP Contamination
Why do we need a cryptographic receipt? If you let an LLM directly commit RTL to your repo, you have no proof of provenance. If the LLM regurgitated copyrighted code, your company is legally liable. The receipt proves that a **human** reviewed the proxy evidence and explicitly authorized the architecture, separating AI generation from human commitment.
