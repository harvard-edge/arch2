## Loading Constraints
Let's load the YAML specifications that act as the physical 'laws' for this tutorial. We are exposing how `arch2_labs` reads these files so it isn't a black box.


````
import yaml
from pathlib import Path
import sys

# NOTE: Our current directory is labs/tutorial, so the root is .parent.parent
root_dir = Path().resolve().parent.parent
yaml_path = root_dir / 'labs' / 'examples' / 'scale_proxy_mirage' / 'configs' / 'candidates.yaml'

with open(yaml_path, 'r') as f:
    spec = yaml.safe_load(f)

print(f"Target Deadline: {spec['verification_environment']['constraints']['deadline_cycles']} cycles")
print(f"Max PE Area Budget: {spec['verification_environment']['constraints']['area_budget_pes']} PEs")
````

## The Human Baseline
We never trust AI blindly. We start with a baseline architecture designed by a human engineer to ensure the verification environment is sane.


````
baseline = next(c for c in spec['candidates'] if c['candidate_id'] == 'balanced_16x16')
print(f"Baseline Rows: {baseline['array_rows']}, Cols: {baseline['array_cols']}")
print(f"Total PEs: {baseline['array_rows'] * baseline['array_cols']}")
if (baseline['array_rows'] * baseline['array_cols']) <= 1024:
    print("✅ Baseline meets area budget.")
````
